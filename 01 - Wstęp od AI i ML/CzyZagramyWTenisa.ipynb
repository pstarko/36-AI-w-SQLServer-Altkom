```python
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree
import matplotlib.pyplot as plt
```

```python
# Dane treningowe
# Pogoda: 0 = Słonecznie, 1 = Pochmurno
# Wilgotność: 0 = Wysoka, 1 = Normalna
# Decyzja: 0 = NIE GRAJ, 1 = GRAJ

X = [
    [0, 0],  # Słonecznie, Wysoka wilgotność
    [0, 1],  # Słonecznie, Normalna wilgotność
    [1, 0],  # Pochmurno, Wysoka wilgotność
    [1, 1],  # Pochmurno, Normalna wilgotność
]

y = [0, 1, 1, 1]  # NIE GRAJ, GRAJ, GRAJ, GRAJ
```

```python
# Tworzenie i trenowanie modelu
clf = DecisionTreeClassifier(random_state=42)
clf.fit(X, y)
```

```python
# Testowanie modelu
print("=== Testowanie modelu ===")
test_cases = [
    ([0, 0], "Słonecznie, Wysoka wilgotność"),
    ([0, 1], "Słonecznie, Normalna wilgotność"),
    ([1, 0], "Pochmurno, Wysoka wilgotność"),
    ([1, 1], "Pochmurno, Normalna wilgotność"),
]

for features, description in test_cases:
    prediction = clf.predict([features])[0]
    decision = "GRAJ!" if prediction == 1 else "NIE GRAJ!"
    print(f"{description}: {decision}")
```

```python
# Wizualizacja drzewa decyzyjnego
plt.figure(figsize=(12, 8))
tree.plot_tree(
    clf,
    feature_names=['Pogoda', 'Wilgotność'],
    class_names=['NIE GRAJ', 'GRAJ'],
    filled=True,
    rounded=True,
    fontsize=10
)
plt.title("Drzewo Decyzyjne - Decyzja o grze w tenisa")
plt.show()
```

```python
print("\n=== Informacje o drzewie ===")
print(f"Głębokość drzewa: {clf.get_depth()}")
print(f"Liczba liści: {clf.get_n_leaves()}")
```
