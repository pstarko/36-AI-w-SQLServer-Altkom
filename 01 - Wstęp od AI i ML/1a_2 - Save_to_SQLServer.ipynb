```python
import pyodbc
import pandas as pd

# Przykładowy DataFrame
df = pd.DataFrame({
    "id": [1, 2, 3],
    "name": ["Ala", "Ola", "Jan"]
})

# Połączenie z SQL Server (dostosuj do swojego serwera)
conn = pyodbc.connect(
    r"DRIVER={ODBC Driver 17 for SQL Server};"
    r"SERVER=STARKOGABINET\SQLSERVER_PYTHON;"   # lub "NAZWA_SERWERA\INSTANCJA"
    r"DATABASE=TEST_JUPYTER;"
    r"UID=python_sql;"      
    r"PWD=oracle123;"
)

cursor = conn.cursor()

# Tworzenie tabeli (przykład)
cursor.execute("""
IF OBJECT_ID('dbo.Users', 'U') IS NOT NULL 
    DROP TABLE dbo.Users;
CREATE TABLE dbo.Users (
    id INT PRIMARY KEY,
    name NVARCHAR(50)
)
""")
conn.commit()

# Wstawianie wartości z DataFrame
for _, row in df.iterrows():
    cursor.execute("INSERT INTO dbo.Users (id, name) VALUES (?, ?)", row["id"], row["name"])
    print(row["name"])

conn.commit()
cursor.close()
conn.close()
```
